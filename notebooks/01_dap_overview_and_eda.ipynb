# Dog Aging Project (DAP): Exploratory Analysis & Codebook Navigation

This notebook provides the foundational environment for analyzing the **Dog Aging Project (DAP)** Curated Data Releases.

### Objectives:
1. Explore the 2024 Curated Data Release (CDR) Codebook schema across instruments.
2. Inspect variable categories (Demographics, Health and Life Experience Survey, Cognitive Assessments, Longevity/Survival).
3. Load and visualize cohort distributions once raw cohorts are downloaded.

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-whitegrid')
pd.set_option('display.max_columns', 100)
print('Environment initialized.')
```

## 1. Load the DAP 2024 Codebook
The codebook catalogs 2,900+ variables spanning the DAP Pack cohorts.

```python
codebook_path = '../data/codebooks/DAP_2024_CODEBOOK_v1.0.csv'
df_codebook = pd.read_csv(codebook_path)
print('Total variables in codebook:', len(df_codebook))
df_codebook.head()
```

```python
# Variable count per DataFile instrument
file_counts = df_codebook['DataFile'].value_counts()
print(file_counts)

plt.figure(figsize=(10, 5))
sns.barplot(x=file_counts.values, y=file_counts.index, palette='viridis')
plt.title('DAP 2024 Variables by DataFile Instrument', fontsize=14)
plt.xlabel('Number of Variables')
plt.ylabel('Instrument / Table')
plt.tight_layout()
plt.show()
```
